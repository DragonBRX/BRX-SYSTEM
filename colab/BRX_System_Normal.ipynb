## 4️⃣ Configurar variáveis de ambiente

```python
from google.colab import userdata
import os

# Configure suas chaves de API nos Secrets do Colab (ícone de chave no menu lateral)
# ou cole diretamente abaixo:
OPENAI_API_KEY = userdata.get('OPENAI_API_KEY') if 'OPENAI_API_KEY' in userdata.keys() else ''
DEEPSEEK_API_KEY = userdata.get('DEEPSEEK_API_KEY') if 'DEEPSEEK_API_KEY' in userdata.keys() else ''
ANTHROPIC_API_KEY = userdata.get('ANTHROPIC_API_KEY') if 'ANTHROPIC_API_KEY' in userdata.keys() else ''
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY') if 'GOOGLE_API_KEY' in userdata.keys() else ''

# Fallback: cole aqui se não usar Secrets
# OPENAI_API_KEY = 'sk-...'

env_content = f"""
NODE_ENV=production
VITE_APP_ID=brx-system-colab
OPENAI_API_KEY={OPENAI_API_KEY}
DEEPSEEK_API_KEY={DEEPSEEK_API_KEY}
ANTHROPIC_API_KEY={ANTHROPIC_API_KEY}
GOOGLE_API_KEY={GOOGLE_API_KEY}
DATABASE_URL=sqlite://./colab.db
"""

with open('.env', 'w') as f:
    f.write(env_content)

print('✅ .env configurado')
```

## 5️⃣ Iniciar o servidor BRX (Backend + Frontend)

```python
import subprocess
import threading
import time

# Compilar e iniciar
!npm run build

def start_server():
    proc = subprocess.Popen(
        ['npm', 'start'],
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True
    )
    for line in proc.stdout:
        print(line, end='')

thread = threading.Thread(target=start_server, daemon=True)
thread.start()
time.sleep(5)
print('🚀 Servidor iniciado na porta padrão')
```

## 6️⃣ Criar túnel público com ngrok (acesso externo)

```python
# Instalar pyngrok
!pip install pyngrok --quiet

from pyngrok import ngrok
from google.colab import userdata

NGROK_TOKEN = userdata.get('NGROK_AUTH_TOKEN') if 'NGROK_AUTH_TOKEN' in userdata.keys() else None
if NGROK_TOKEN:
    ngrok.set_auth_token(NGROK_TOKEN)

public_url = ngrok.connect(3000)
print(f'🌐 URL Pública: {public_url}')
print('Abra o link acima para acessar o BRX SYSTEM no navegador!')
```
