## 2️⃣ Iniciar Ollama em background

```python
import subprocess, time
# Iniciar servidor Ollama
ollama_proc = subprocess.Popen(['ollama', 'serve'], stdout=subprocess.PIPE, stderr=subprocess.PIPE)
time.sleep(5)
print('🦙 Ollama rodando na porta 11434')
```

## 3️⃣ Baixar modelo leve (ex: Phi-2 ou TinyLlama)

```python
!ollama pull phi
# Alternativas leves: tinyllama, qwen:0.5b, gemma:2b
print('✅ Modelo local pronto')
```

## 4️⃣ Clonar e preparar BRX SYSTEM

```python
!git clone https://github.com/DragonBRX/BRX-SYSTEM.git && %cd BRX-SYSTEM
!npm install --silent
```

## 5️⃣ Configurar ambiente local

```python
env = """
NODE_ENV=production
VITE_APP_ID=brx-local-colab
OLLAMA_URL=http://localhost:11434
BRX_MODE=local
BRX_LOCAL_MODEL=phi
DATABASE_URL=sqlite://./local.db
"""
with open('.env', 'w') as f:
    f.write(env)
print('✅ Configurado para modelo local via Ollama')
```

## 6️⃣ Build & Start

```python
import subprocess, threading, time
!npm run build
def run():
    p = subprocess.Popen(['npm', 'start'], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in p.stdout:
        print(line, end='')
threading.Thread(target=run, daemon=True).start()
time.sleep(5)
print('🚀 BRX Local iniciado!')
```

## 7️⃣ Túnel público

```python
!pip install pyngrok --quiet
from pyngrok import ngrok
from google.colab import userdata
token = userdata.get('NGROK_AUTH_TOKEN') if 'NGROK_AUTH_TOKEN' in userdata.keys() else None
if token:
    ngrok.set_auth_token(token)
url = ngrok.connect(3000)
print(f'🌐 Acesse seu BRX Local: {url}')
```
