## 4️⃣ Configurar modo multi-agentes + API Key

```python
from google.colab import userdata

# Recomendamos Google Gemini (tier gratuito generoso) ou DeepSeek
GOOGLE_API_KEY = userdata.get('GOOGLE_API_KEY') if 'GOOGLE_API_KEY' in userdata.keys() else ''
DEEPSEEK_API_KEY = userdata.get('DEEPSEEK_API_KEY') if 'DEEPSEEK_API_KEY' in userdata.keys() else ''

env = f"""
NODE_ENV=production
VITE_APP_ID=brx-multiagent-colab
GOOGLE_API_KEY={GOOGLE_API_KEY}
DEEPSEEK_API_KEY={DEEPSEEK_API_KEY}
BRX_MODE=multi_agent
BRX_MAX_AGENTS=5
BRX_SHARED_MEMORY=true
DATABASE_URL=sqlite://./multiagent.db
"""
with open('.env', 'w') as f:
    f.write(env)
print('✅ Modo Multi-Agente configurado (até 5 agentes com memória compartilhada)')
```

## 5️⃣ Build & Start

```python
import subprocess, threading, time
!npm run build
def run():
    p = subprocess.Popen(['npm', 'start'], stdout=subprocess.PIPE, stderr=subprocess.STDOUT, text=True)
    for line in p.stdout:
        print(line, end='')
threading.Thread(target=run, daemon=True).start()
time.sleep(5)
print('🚀 Multi-Agente iniciado')
```

## 6️⃣ Túnel público (ngrok)

```python
!pip install pyngrok --quiet
from pyngrok import ngrok
from google.colab import userdata
token = userdata.get('NGROK_AUTH_TOKEN') if 'NGROK_AUTH_TOKEN' in userdata.keys() else None
if token:
    ngrok.set_auth_token(token)
url = ngrok.connect(3000)
print(f'🌐 Acesse: {url}')
```
